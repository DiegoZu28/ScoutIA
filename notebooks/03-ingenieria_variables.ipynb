# Ingeniería de variables para el modelado de `valor_eur`

Partimos de `data/processed/fbref_tm_eda.parquet`, el dataset ya tratado en el EDA (`02-eda.ipynb`): sin duplicados de tipo, con nulos estructurales imputados (y sus flags), atípicos conservados (son la señal de mercado) y con columnas auxiliares ya normalizadas (`_rs`) o log-transformadas (`log1p_`).

**Objetivo de este notebook:** dejar `df` listo como matriz de variables para predecir `valor_eur` (o `log_valor_eur`, su versión log-transformada). Para ello:

1. Se corrigen errores de cruce FBref↔Transfermarkt (filas donde las estadísticas de un jugador quedaron pegadas al valor de mercado de otro).
2. Se eliminan identificadores, columnas de auditoría de calidad de datos y una variable con **fuga de información**.
3. Se eliminan duplicados exactos y variables derivadas (combinaciones lineales de otras que ya se conservan).
4. Se resuelve la redundancia entre columnas "en bruto" y sus versiones ya tratadas en el EDA (`_rs` robust-scaled, `log1p_`), quedándonos con una sola representación por variable.
5. Se construyen **variables categóricas (dummies)** para posición, competición y equipo.
6. Se documenta, variable por variable, el tratamiento aplicado y el motivo.

Cada sección de código va acompañada de una celda markdown que explica **qué se hace y por qué**, para que el tratamiento quede trazable.

## 1. Diagnóstico previo: qué trae `df` heredado del EDA

Antes de decidir qué eliminar o transformar, revisamos tipos de dato, nulos restantes y cardinalidad de las categóricas, y detectamos columnas duplicadas o derivadas por combinación lineal de otras (evidencian redundancia exacta, no solo correlación alta).

```python
print(f"Shape inicial: {df.shape}")
print()
print("Nulos restantes:")
print(df.isna().sum().pipe(lambda s: s[s > 0]))
print()
print("Cardinalidad de columnas categóricas / booleanas:")
for c in df.columns:
    if df[c].dtype.name in ("category", "object", "bool"):
        print(f"  {c:26s} {df[c].nunique():>4d} categorías")
```

## 2. Corrección de cruce y eliminación de identificadores/auditoría

### 2.1 Filas con cruce FBref↔Transfermarkt ambiguo

Al revisar los residuos de un Random Forest en `04-modelado.ipynb`, las predicciones para "Rodri" eran inconsistentes entre filas que decían tener el mismo `valor_eur`. La causa está en `01-unificacion.ipynb`: el cruce FBref↔Transfermarkt se hace por nombre normalizado (con año de nacimiento como *fallback*), y el merge final (`fb.merge(tmu, on=['player_id','saison_id'], how='left', validate='many_to_one')`) permite que **más de una fila de FBref** —jugadores distintos que comparten nombre normalizado— quede pegada al mismo `player_id` (y por lo tanto al mismo `valor_eur`) de Transfermarkt en la misma temporada.

Esto afecta a **735 grupos `(player_id, saison_id)`** (1.473 filas, ~10.6% del dataset) con más de un `Squad` distinto para el mismo `player_id` en la misma temporada — es decir, estadísticas de un jugador equivocado pegadas al valor de mercado de otro. La columna `club_coincide` (¿el `Squad` de FBref coincide con el `club` de Transfermarkt de esa temporada?), calculada en el cruce, permite distinguir la fila correcta:

- **660 grupos**: exactamente una fila tiene `club_coincide=True` → se conserva esa fila (la correcta) y se descartan las demás del grupo.
- **75 grupos**: ninguna fila coincide con el club de Transfermarkt → no hay forma confiable de saber cuál (si alguna) es la correcta, se descarta el grupo completo.

Este tratamiento se aplica **antes** de descartar `club_coincide` como columna de auditoría (sección 2.2), porque hace falta para esta limpieza.

### 2.2 Identificadores, columnas de auditoría y fuga de información

| Columna | Tratamiento | Motivo |
|---|---|---|
| `Rk` | Eliminar | Índice de scraping sin significado, no es un atributo del jugador. |
| `nom` | Eliminar | Nombre normalizado en minúsculas, usado solo como clave de cruce FBref↔Transfermarkt (redundante con `Player`). |
| `pid_fb` | Eliminar | Duplica `player_id` en el 99.8% de los casos; es un campo de auditoría del cruce, no un atributo. |
| `anio_nac` | Eliminar | Duplica `Born`/`nacimiento`, y ambos a su vez son redundantes con `Age`. |
| `Born` | Eliminar | Redundante con `Age` (año de nacimiento vs. edad calculada); se conserva `Age`. |
| `nacimiento` | Eliminar | La información relevante (edad) ya queda resumida en `Age`; el día/mes exacto de nacimiento no aporta valor de mercado adicional. |
| `club` | Eliminar | Según el EDA, es una *instantánea* del club actual en Transfermarkt, no un histórico por temporada (puede no coincidir con el club real en temporadas antiguas). Se usa `Squad`, que sí es específico de cada temporada de FBref. |
| `Season` | Eliminar | Redundante con `saison_id` (mismo valor pero en formato texto "2021-2022" vs. entero 2021); se conserva la versión numérica. |
| `flag_cruce_verificado` | Eliminar | Flag de auditoría del cruce de fuentes (si `pid_fb` existía o no) — describe la calidad del dato, no una característica del jugador que explique su valor de mercado. |
| `club_coincide` | Eliminar (ya usada en 2.1) | Igual que el anterior: flag de calidad de datos, no un atributo del jugador — pero antes de eliminarla se usó para resolver los cruces ambiguos de la sección 2.1. |
| `n_metodos_atipico` | **Eliminar (fuga de información)** | En el EDA se calculó ejecutando IsolationForest / LOF / EllipticEnvelope / Mahalanobis usando **`valor_eur` como una de las variables de entrada**. Usarlo como *feature* filtraría directamente información del objetivo al modelo. |

`player_id` y `Player` se conservan, pero **solo como identificadores** (no como *features* de entrada al modelo): permiten trazabilidad y, como señala el EDA, dividir train/test por `player_id` para evitar que el mismo jugador aparezca en ambos conjuntos.

```python
key_cols = ["player_id", "saison_id"]
n_squads = df.groupby(key_cols)["Squad"].transform("nunique")
ambiguos = n_squads > 1
n_true_grupo = df.groupby(key_cols)["club_coincide"].transform("sum")

n_grupos_ambiguos = df.loc[ambiguos, key_cols].drop_duplicates().shape[0]
n_grupos_resueltos = df.loc[ambiguos & (n_true_grupo == 1), key_cols].drop_duplicates().shape[0]
n_grupos_sin_confianza = df.loc[ambiguos & (n_true_grupo == 0), key_cols].drop_duplicates().shape[0]

print(f"Grupos (player_id, saison_id) con Squad ambiguo: {n_grupos_ambiguos}")
print(f"  - resueltos via club_coincide (queda 1 fila correcta): {n_grupos_resueltos}")
print(f"  - sin ninguna fila confiable, se descarta el grupo completo: {n_grupos_sin_confianza}")

descartar = (
    (ambiguos & (n_true_grupo == 1) & (~df["club_coincide"]))
    | (ambiguos & (n_true_grupo == 0))
)

print(f"\nFilas descartadas: {descartar.sum()} de {df.shape[0]}")
df = df[~descartar].reset_index(drop=True)
print(f"Shape actual: {df.shape}")
```

```python
cols_identificadores_auditoria = [
    "Rk", "nom", "pid_fb", "anio_nac", "Born", "nacimiento", "club", "Season",
    "flag_cruce_verificado", "club_coincide",
]
cols_leakage = ["n_metodos_atipico"]

df = df.drop(columns=cols_identificadores_auditoria + cols_leakage)
print(f"Eliminadas {len(cols_identificadores_auditoria)} columnas de identificadores/auditoría "
      f"y {len(cols_leakage)} columna con fuga de información.")
print(f"Shape actual: {df.shape}")
```

## 3. Duplicados exactos y variables derivadas (combinación lineal)

FBref repite algunas métricas en varias tablas de su web (todas fueron scrapeadas y quedaron en el dataset con nombres distintos). Verificamos que son **idénticas fila a fila** antes de eliminarlas, y lo mismo para columnas que son resultado de sumar/restar otras dos que también se conservan (colinealidad perfecta, no solo alta correlación):

| Columna a eliminar | Motivo (idéntica o derivada de...) |
|---|---|
| `Standard_Gls` | = `Performance_Gls` (100% igual) |
| `Standard_PK` | = `Performance_PK` (100% igual) |
| `Standard_PKatt` | = `Performance_PKatt` (100% igual) |
| `90s` | = `Playing Time_90s` (100% igual) |
| `Starts_Starts` | = `Playing Time_Starts` (100% igual) |
| `Performance_G+A` | = `Performance_Gls` + `Performance_Ast` |
| `Performance_G-PK` | = `Performance_Gls` − `Performance_PK` |
| `Per 90 Minutes_G+A` | = `Per 90 Minutes_Gls` + `Per 90 Minutes_Ast` |

| `Per 90 Minutes_G+A-PK` | = versión por-90 de `Performance_Gls` + `Performance_Ast` − `Performance_PK` |
| `Team Success_+/-` | = `Team Success_onG` − `Team Success_onGA` |
| `Team Success_+/-90` | = `Team Success_+/-` normalizado por `90s` |

En todos los casos se conserva **una sola** de las columnas equivalentes (la de nombre más simple/estándar) y se descartan las derivadas, ya que no aportan información nueva y solo inflarían la multicolinealidad.

```python
# Verificación de igualdad exacta antes de eliminar
dup_pairs = [
    ("Standard_Gls", "Performance_Gls"),
    ("Standard_PK", "Performance_PK"),
    ("Standard_PKatt", "Performance_PKatt"),
    ("90s", "Playing Time_90s"),
    ("Starts_Starts", "Playing Time_Starts"),
]
for a, b in dup_pairs:
    print(f"{a:18s} == {b:22s} -> {(df[a] == df[b]).mean():.3f} de las filas iguales")

cols_duplicadas = [a for a, _ in dup_pairs]
cols_derivadas = [
    "Performance_G+A", "Performance_G-PK",
    "Per 90 Minutes_G+A", "Per 90 Minutes_G-PK", "Per 90 Minutes_G+A-PK",
    "Team Success_+/-", "Team Success_+/-90",
]

df = df.drop(columns=cols_duplicadas + cols_derivadas)
print(f"\nEliminadas {len(cols_duplicadas)} columnas duplicadas y {len(cols_derivadas)} derivadas.")
print(f"Shape actual: {df.shape}")
```

## 5. Relativizar variables ofensivas y defensivas por posición

`Performance_Gls`, `Performance_Ast`, `Standard_Sh` (producción ofensiva) y `Performance_TklW`, `Performance_Int` (involucramiento defensivo) significan cosas muy distintas según la posición: 5 goles es excepcional para un central pero mediocre para un delantero, y el argumento espejo aplica a entradas/intercepciones (normal para un mediocentro defensivo, raro para un delantero. Se reemplaza el escalado global de estas 5 variables por un `RobustScaler` ajustado **dentro de cada `posicion_tm`**, para que cada valor se compare contra jugadores de su misma posición y no contra todo el dataset. Tiene que calcularse acá, antes de que la sección 8 convierta `posicion_tm` en dummies y la descarte.

**Nota técnica:** para porteros estas 5 estadísticas son prácticamente siempre 0 (mediana e IQR = 0 dentro del grupo). `RobustScaler` ya maneja ese caso de forma segura (deja `scale_ = 1` en vez de dividir por 0), así que no hace falta un tratamiento especial aparte -- simplemente quedan en 0 tras el escalado, correcto para un grupo sin varianza.

```python
from sklearn.preprocessing import RobustScaler

STATS_POR_POSICION = ["Performance_Gls", "Performance_Ast", "Standard_Sh", "Performance_TklW", "Performance_Int"]

partes = []
for pos, sub in df.groupby("posicion_tm", observed=True):
    rs = RobustScaler()
    escalado = rs.fit_transform(sub[STATS_POR_POSICION])
    partes.append(pd.DataFrame(escalado, columns=[f"{c}_rs" for c in STATS_POR_POSICION], index=sub.index))

df_pos_rs = pd.concat(partes).sort_index()
df[df_pos_rs.columns] = df_pos_rs  # sobreescribe Gls/Ast/Sh (antes globales) y crea TklW/Int (nuevas)

# Cada posición queda centrada en su propia mediana (~0), a diferencia del escalado global
print("Mediana de Performance_Gls_rs por posición (debería rondar 0 en todas):")
print(df.groupby("posicion_tm", observed=True)["Performance_Gls_rs"].median().round(2))
```

```python
# Performance_Gls/Ast/Standard_Sh (crudos) los elimina la sección 6 más abajo (ya
# reemplazados por su _rs, ahora por posición). TklW/Int no tenían tratamiento previo:
df = df.drop(columns=["Performance_TklW", "Performance_Int"])
print(f"Shape actual: {df.shape}")
```

## 6. Consolidar variables "en bruto" con su versión ya tratada en el EDA

El EDA (`02-eda.ipynb`) ya generó, para las variables más sesgadas, una versión transformada:

- **`_rs`** (RobustScaler: `(x − mediana) / IQR`) para `Age`, `altura_m`, `Playing Time_Min`, `Performance_Gls`, `Performance_Ast`, `Standard_Sh`, `Performance_CrdY`, `Team Success_PPM`.
- **`log1p_`** para variables de conteo muy asimétricas: `Performance_Gls`, `Performance_Ast`, `Standard_Sh`, `Standard_SoT`, `Performance_Crs`, `Performance_Fls`.

Mantener a la vez la columna original, su `log1p_` y su `_rs` sería usar **la misma señal tres veces** en distintas escalas. Se define una única representación final por variable:

| Variable base | Representación final elegida | Se descartan |
|---|---|---|
| `Age` | `Age_rs` | `Age` |
| `altura_m` | `altura_m_rs` | `altura_m` |
| `Playing Time_Min` | `Playing Time_Min_rs` | `Playing Time_Min` |
| `Performance_CrdY` | `Performance_CrdY_rs` | `Performance_CrdY` |
| `Team Success_PPM` | `Team Success_PPM_rs` | `Team Success_PPM` |
| `Performance_Gls` | `Performance_Gls_rs` | `Performance_Gls`, `log1p_Performance_Gls` |
| `Performance_Ast` | `Performance_Ast_rs` | `Performance_Ast`, `log1p_Performance_Ast` |
| `Standard_Sh` | `Standard_Sh_rs` | `Standard_Sh`, `log1p_Standard_Sh` |
| `Standard_SoT` | `log1p_Standard_SoT` | `Standard_SoT` |
| `Performance_Crs` | `log1p_Performance_Crs` | `Performance_Crs` |
| `Performance_Fls` | `log1p_Performance_Fls` | `Performance_Fls` |

Criterio: para las variables con `_rs` disponible se usa el escalado robusto (mediana/IQR), que ya vimos en el EDA que preserva mejor la forma de la distribución sin comprimir la señal de los jugadores más usados/productivos. Para las que solo cuentan con `log1p_` se usa esa versión, que corrige directamente la asimetría de la variable de conteo.

```python
cols_reemplazadas_por_rs = [
    "Age", "altura_m", "Playing Time_Min", "Performance_CrdY", "Team Success_PPM",
]
cols_reemplazadas_por_rs_y_log1p = ["Performance_Gls", "Performance_Ast", "Standard_Sh"]
cols_log1p_duplicado = [f"log1p_{c}" for c in cols_reemplazadas_por_rs_y_log1p]
cols_reemplazadas_por_log1p = ["Standard_SoT", "Performance_Crs", "Performance_Fls"]

df = df.drop(columns=(
    cols_reemplazadas_por_rs
    + cols_reemplazadas_por_rs_y_log1p
    + cols_log1p_duplicado
    + cols_reemplazadas_por_log1p
))
print(f"Shape actual: {df.shape}")
```

## 8. Variables categóricas: posición, competición y equipo

Se evalúan tres categóricas: `posicion_tm`, `Comp` y `Squad`. Las dos primeras tienen cardinalidad baja (13 y 5 categorías) y se codifican con one-hot (`pd.get_dummies`, `drop_first=True` para evitar la *dummy variable trap* en modelos lineales/regularizados; en árboles no supone pérdida de información). `Squad`, con 130 categorías, recibe un tratamiento distinto (ver 8.3).

### 8.1 Posición (`posicion_tm`)
`Pos` (FBref) y `posicion_tm` (Transfermarkt) son consistentes entre sí (ver heatmap del EDA), pero `posicion_tm` es más granular (13 categorías puras: `Centre-Back`, `Left Winger`, `Defensive Midfield`, …) mientras que `Pos` mezcla combinaciones (`DF,MF`, `MF,FW`) para jugadores multiposición. Se usa **`posicion_tm`** como fuente de la dummy de posición y se elimina `Pos` para no duplicar la misma información con una codificación más ambigua.

### 7.2 Competición (`Comp`)
Solo 5 categorías (las 5 grandes ligas). El EDA mostró diferencias claras en la mediana de `log_valor_eur` por liga (Premier League y La Liga por encima del resto), por lo que aporta señal directa de "nivel de mercado" de cada liga. Se codifica completa en dummies.

### 7.3 Equipo (`Squad`): por qué no one-hot

`Squad` tiene 130 categorías. Un one-hot completo generaría ~129 columnas casi siempre dispersas, muchas con apenas 1-2 observaciones (equipos que solo pasaron una temporada por las 5 grandes ligas). Agrupar en un top-N + `"Otro_equipo"` (como se hacía en una versión anterior de este notebook) reduce la dimensionalidad, pero colapsa la mayoría de los equipos en una sola categoría, perdiendo toda la señal individual de clubes medianos (p. ej. Betis, Villarreal, Eintracht Frankfurt) que sí tienen un efecto propio y distinto sobre el valor de mercado.

Opciones evaluadas para reemplazar el one-hot:

| Método | Cómo funciona | Ventajas | Problema para este caso |
|---|---|---|---|
| **Label Encoding** | Cada equipo → un entero arbitrario (0, 1, 2, …) | Trivial, una sola columna | Introduce una relación de orden falsa entre equipos que no existe (`Zaragoza=45` no es "más" que `Betis=12`); modelos lineales la tomarían literalmente, y hasta en árboles el orden arbitrario del encoder puede sesgar los splits |
| **Frequency / Count Encoding** | Cada equipo → nº de filas jugador-temporada que tiene en el dataset | Una sola columna, no usa el target (cero riesgo de fuga), robusta | Mide *tamaño de plantilla / permanencia en el dataset*, no *prestigio o valor de mercado*: dos equipos con frecuencia similar pueden tener valores de mercado muy distintos |
| **Target Encoding (mean encoding suavizado)** | Cada equipo → media de `log_valor_eur` para ese equipo, regularizada hacia la media global según su frecuencia (así los equipos con pocas observaciones no quedan con una media ruidosa) | Captura directamente la señal que motivaba el one-hot (los clubes "grandes" inflan el valor de sus jugadores), conserva las 130 categorías en una sola columna | Usa el target: si se calcula sobre todo el dataset **antes** de separar train/test, filtra información del test hacia las features de train |

**Decisión:** Target Encoding es el que mejor sustituye la señal que buscaba capturar el one-hot, sin la explosión dimensional del one-hot ni la pérdida de información del agrupamiento top-N. Para no introducir fuga de información, **no se calcula en este notebook**: aquí `Squad` se deja como columna categórica sin codificar, y el target encoding se ajusta en `04-modelado.ipynb` **solo con los datos de train**, después del split por `player_id` (con `sklearn.preprocessing.TargetEncoder`, que además aplica *cross-fitting* interno para no sobreajustar dentro del propio train). Es el mismo criterio de rigor que ya se aplicó en la sección 2 con `n_metodos_atipico`.

```python
# 7.1 Posición: dummies desde posicion_tm, se elimina Pos (redundante, menos granular)
df = df.drop(columns=["Pos"])
dummies_pos = pd.get_dummies(df["posicion_tm"], prefix="pos", drop_first=True)

# 7.2 Competición: dummies desde Comp
dummies_comp = pd.get_dummies(df["Comp"], prefix="liga", drop_first=True)

print(f"Dummies de posición: {dummies_pos.shape[1]}")
print(f"Dummies de competición: {dummies_comp.shape[1]}")

df = pd.concat([df, dummies_pos, dummies_comp], axis=1)
df = df.drop(columns=["posicion_tm", "Comp"])

# 7.3 Equipo: se deja `Squad` como categórica sin codificar (130 categorías).
# El target encoding se ajusta en 04-modelado.ipynb solo con el train, tras el split.
df["Squad"] = df["Squad"].astype(str)
print(f"\nSquad se conserva sin codificar: {df['Squad'].nunique()} categorías (pendiente de target encoding en 04-modelado.ipynb)")

print(f"\nShape actual: {df.shape}")
```

## 9. Vencimiento de contrato: representación final

`contrato_vence` es una fecha cruda -- no es una *feature* utilizable tal cual (ni los modelos lineales ni los de árbol de este proyecto aceptan `datetime`). La señal numérica ya vive en `dias_contrato_restante` (derivada en `02-eda.ipynb`), así que se elimina la fecha cruda y **solo falta decidir qué hacer con su 43% de nulos**, ya que `LinearRegression` y `RandomForestRegressor` no toleran `NaN` (XGBoost sí, pero hay que dejar una sola matriz de *features* válida para los cinco modelos de `04-modelado.ipynb`).

**Tratamiento:** se imputa con la mediana global (elección neutra: no asume que el jugador con contrato desconocido tiene *más* o *menos* tiempo restante que el resto) y se conserva `flag_contrato_conocido` (ya creada) para que el modelo pueda aprender si el hecho de no tener el dato es, en sí mismo, informativo.

```python
mediana_dias = df["dias_contrato_restante"].median()
df["dias_contrato_restante"] = df["dias_contrato_restante"].fillna(mediana_dias)
df = df.drop(columns=["contrato_vence"])

print(f"Mediana usada para imputar: {mediana_dias:.0f} días")
print(f"Nulos restantes en dias_contrato_restante: {df['dias_contrato_restante'].isna().sum()}")
print(f"Shape actual: {df.shape}")
```
