# Ingeniería de variables para predecir `valor_eur`

Partimos del dataset que dejó el EDA (`02-eda.ipynb`): sin duplicados raros, nulos ya imputados con sus flags, atípicos conservados a propósito, y algunas columnas ya normalizadas (`_rs`, `log1p_`).

Acá dejamos `df` listo como matriz de variables para el modelo: corregimos un bug de cruce que encontramos después, sacamos identificadores/columnas de auditoría y una fuga de información, eliminamos duplicados y derivadas, decidimos una sola representación por variable, y armamos las dummies de posición/liga/equipo. Cada paso de código va con su celda explicando el porqué.

## 1. Qué trae `df` heredado del EDA

Antes de tocar nada, miramos tipos de dato, nulos que quedan, y qué columnas están duplicadas o son combinación lineal de otras -- redundancia exacta, no solo correlación alta.

```python
print(f"Shape inicial: {df.shape}")
print()
print("Nulos restantes:")
print(df.isna().sum().pipe(lambda s: s[s > 0]))
print()
print("Cardinalidad de columnas categóricas / booleanas:")
for c in df.columns:
    if df[c].dtype.name in ("category", "object", "bool"):
        print(f"  {c:26s} {df[c].nunique():>4d} categorías")
```

## 2. Arreglando un bug de cruce, y sacando columnas que no sirven de feature

### 2.1 Cuando dos jugadores distintos quedan pegados

Esto lo encontramos mirando los residuos de un Random Forest en `04-modelado.ipynb`: las predicciones para "Rodri" no tenían sentido entre filas que se supone tenían el mismo `valor_eur`. La causa está en el cruce de `01-unificacion.ipynb` -- se hace por nombre, y a veces dos jugadores distintos con el mismo nombre normalizado terminan compartiendo el mismo `player_id` de Transfermarkt (y su valor de mercado).

Afecta a 735 grupos jugador-temporada (~10.6% del dataset). Para resolverlo usamos `club_coincide` (¿el `Squad` de FBref coincide con el club de Transfermarkt?): si una sola fila del grupo coincide, esa es la buena y tiramos el resto; si ninguna coincide, no hay forma confiable de saber cuál es la correcta y se descarta el grupo entero.

**Vuelta después de entregar el proyecto:** un jugador cedido a otro club también genera dos `Squad` en la misma temporada, pero ahí NO es una colisión -- es la misma persona. `club_coincide` no distinguía esto, y como el club de Transfermarkt siempre es el dueño de la ficha (nunca el que recibe la cesión), terminaba tirando la fila de la cesión. Nos dimos cuenta con Endrick (cedido todo 2025-2026 en el Olympique Lyon, 16 partidos y 5 goles) -- se quedaba solo con los 12 minutos que jugó en el Real Madrid después de volver. Arreglamos esto con `scripts/detectar_prestamos.py`, que lee directamente las insignias de cesión de las plantillas de Transfermarkt y, cuando hay evidencia de que fue cedido, prioriza esa fila por sobre `club_coincide`.

### 2.2 Identificadores y columnas de auditoría

| Columna | Por qué se va |
|---|---|
| `Rk` | Índice del scraping, no dice nada del jugador |
| `nom`, `pid_fb`, `anio_nac` | Solo sirvieron para el cruce de fuentes, ya cumplieron su función |
| `Born`, `nacimiento` | Redundantes con `Age` |
| `club` | Es una foto del club actual, no un histórico -- usamos `Squad` |
| `Season` | Redundante con `saison_id` |
| `flag_cruce_verificado`, `club_coincide` | Auditoría del cruce, no un atributo del jugador |
| `n_metodos_atipico` | **Fuga de información** -- se calculó usando `valor_eur` como input en el EDA |

`player_id` y `Player` se quedan, pero solo como identificadores (para trazabilidad y para separar train/test por jugador), nunca como feature del modelo.

```python
import re
import unicodedata


def normalizar_club(s):
    s = unicodedata.normalize("NFKD", str(s))
    s = "".join(c for c in s if not unicodedata.combining(c))
    s = s.replace("đ", "d").replace("Đ", "D").replace("ł", "l").replace("ø", "o")
    s = re.sub(r"[^a-z ]", "", s.lower())
    return re.sub(r"\s+", " ", s).strip()


def coincide(a, b):
    a, b = normalizar_club(a), normalizar_club(b)
    return (a in b) or (b in a)


# Evidencia de cesión (insignias "On loan from X" / "Returned after loan spell with X" de
# las plantillas de Transfermarkt, ver scripts/detectar_prestamos.py). club_coincide por sí
# solo siempre prefiere al club DUEÑO de la ficha -- para un jugador cedido, eso descarta
# sistemáticamente la fila de la cesión (a veces su temporada entera), aunque ambas filas
# sean la MISMA persona, no una colisión de nombre. Caso encontrado por el usuario:
# Endrick, cedido en Olympique Lyon toda 2025-2026 (16 partidos, 5 goles) -- club_coincide
# se quedaba con la fila de Real Madrid, sus 12 minutos post-regreso de la cesión.
prestamos = pd.read_csv("../data/interim/prestamos.csv", dtype={"player_id": str})
club_prestamo_de = dict(zip(zip(prestamos["player_id"], prestamos["saison_id"]), prestamos["club_prestamo"]))
club_prestamo = [club_prestamo_de.get((pid, sid)) for pid, sid in zip(df["player_id"], df["saison_id"])]
df["_es_fila_prestamo"] = [
    cp is not None and coincide(sq, cp) for sq, cp in zip(df["Squad"], club_prestamo)
]

key_cols = ["player_id", "saison_id"]
n_squads = df.groupby(key_cols)["Squad"].transform("nunique")
ambiguos = n_squads > 1
tiene_prestamo = df.groupby(key_cols)["_es_fila_prestamo"].transform("any")
n_true_grupo = df.groupby(key_cols)["club_coincide"].transform("sum")

n_grupos_ambiguos = df.loc[ambiguos, key_cols].drop_duplicates().shape[0]
n_grupos_prestamo = df.loc[ambiguos & tiene_prestamo, key_cols].drop_duplicates().shape[0]
n_grupos_resueltos = df.loc[ambiguos & ~tiene_prestamo & (n_true_grupo == 1), key_cols].drop_duplicates().shape[0]
n_grupos_sin_confianza = df.loc[ambiguos & ~tiene_prestamo & (n_true_grupo == 0), key_cols].drop_duplicates().shape[0]

print(f"Grupos (player_id, saison_id) con Squad ambiguo: {n_grupos_ambiguos}")
print(f"  - resueltos via evidencia de cesión (queda la fila del club donde jugó): {n_grupos_prestamo}")
print(f"  - resueltos via club_coincide (queda 1 fila correcta): {n_grupos_resueltos}")
print(f"  - sin ninguna fila confiable, se descarta el grupo completo: {n_grupos_sin_confianza}")

descartar_club_coincide = (
    (ambiguos & (n_true_grupo == 1) & (~df["club_coincide"]))
    | (ambiguos & (n_true_grupo == 0))
)
descartar_prestamo = ambiguos & tiene_prestamo & (~df["_es_fila_prestamo"])
descartar = np.where(tiene_prestamo, descartar_prestamo, descartar_club_coincide)

print(f"\nFilas descartadas: {descartar.sum()} de {df.shape[0]}")
df = df[~descartar].reset_index(drop=True)
df = df.drop(columns=["_es_fila_prestamo"])
print(f"Shape actual: {df.shape}")
```

```python
cols_identificadores_auditoria = [
    "Rk", "nom", "pid_fb", "anio_nac", "Born", "nacimiento", "club", "Season",
    "flag_cruce_verificado", "club_coincide",
]
cols_leakage = ["n_metodos_atipico"]

df = df.drop(columns=cols_identificadores_auditoria + cols_leakage)
print(f"Eliminadas {len(cols_identificadores_auditoria)} columnas de identificadores/auditoría "
      f"y {len(cols_leakage)} columna con fuga de información.")
print(f"Shape actual: {df.shape}")
```

## 3. Duplicados exactos y variables derivadas

FBref repite algunas métricas con nombres distintos en sus tablas. Confirmamos que son idénticas (o suma/resta exacta de otras que ya tenemos) antes de tirarlas -- no aportan nada nuevo, solo multicolinealidad:

| Columna que se va | Es igual a... |
|---|---|
| `Standard_Gls` | `Performance_Gls` |
| `Standard_PK` / `Standard_PKatt` | `Performance_PK` / `Performance_PKatt` |
| `90s` | `Playing Time_90s` |
| `Starts_Starts` | `Playing Time_Starts` |
| `Performance_G+A` | `Gls` + `Ast` |
| `Performance_G-PK` | `Gls` − `PK` |
| `Per 90 Minutes_G+A`, `G-PK`, `G+A-PK` | versiones por-90 de lo mismo |
| `Team Success_+/-` y `+/-90` | `onG` − `onGA`, normalizado |

Nos quedamos con una sola versión de cada una.

```python
# Verificación de igualdad exacta antes de eliminar
dup_pairs = [
    ("Standard_Gls", "Performance_Gls"),
    ("Standard_PK", "Performance_PK"),
    ("Standard_PKatt", "Performance_PKatt"),
    ("90s", "Playing Time_90s"),
    ("Starts_Starts", "Playing Time_Starts"),
]
for a, b in dup_pairs:
    print(f"{a:18s} == {b:22s} -> {(df[a] == df[b]).mean():.3f} de las filas iguales")

cols_duplicadas = [a for a, _ in dup_pairs]
cols_derivadas = [
    "Performance_G+A", "Performance_G-PK",
    "Per 90 Minutes_G+A", "Per 90 Minutes_G-PK", "Per 90 Minutes_G+A-PK",
    "Team Success_+/-", "Team Success_+/-90",
]

df = df.drop(columns=cols_duplicadas + cols_derivadas)
print(f"\nEliminadas {len(cols_duplicadas)} columnas duplicadas y {len(cols_derivadas)} derivadas.")
print(f"Shape actual: {df.shape}")
```

## 5. Relativizar por posición las variables ofensivas y defensivas

5 goles es una temporada excepcional para un central pero mediocre para un delantero -- lo mismo al revés con entradas/intercepciones. Por eso re-escalamos `Gls`, `Ast`, `Sh`, `TklW` e `Int` con un `RobustScaler` calculado **dentro de cada posición**, no contra todo el dataset. Esto hay que hacerlo antes de que la sección 8 convierta `posicion_tm` en dummies.

(Para porteros estas estadísticas son casi siempre 0 -- `RobustScaler` lo maneja bien solo, no hace falta nada especial.)

```python
from sklearn.preprocessing import RobustScaler

STATS_POR_POSICION = ["Performance_Gls", "Performance_Ast", "Standard_Sh", "Performance_TklW", "Performance_Int"]

partes = []
for pos, sub in df.groupby("posicion_tm", observed=True):
    rs = RobustScaler()
    escalado = rs.fit_transform(sub[STATS_POR_POSICION])
    partes.append(pd.DataFrame(escalado, columns=[f"{c}_rs" for c in STATS_POR_POSICION], index=sub.index))

df_pos_rs = pd.concat(partes).sort_index()
df[df_pos_rs.columns] = df_pos_rs  # sobreescribe Gls/Ast/Sh (antes globales) y crea TklW/Int (nuevas)

# Cada posición queda centrada en su propia mediana (~0), a diferencia del escalado global
print("Mediana de Performance_Gls_rs por posición (debería rondar 0 en todas):")
print(df.groupby("posicion_tm", observed=True)["Performance_Gls_rs"].median().round(2))
```

```python
# Performance_Gls/Ast/Standard_Sh (crudos) los elimina la sección 6 más abajo (ya
# reemplazados por su _rs, ahora por posición). TklW/Int no tenían tratamiento previo:
df = df.drop(columns=["Performance_TklW", "Performance_Int"])
print(f"Shape actual: {df.shape}")
```

## 6. Una sola versión final por variable

El EDA ya dejó, para las variables más sesgadas, una versión `_rs` (robust-scaled) y/o `log1p_`. Mantener la original + las dos transformadas sería usar la misma señal tres veces. Elegimos una representación final por variable:

| Variable | Nos quedamos con | Se van |
|---|---|---|
| `Age`, `altura_m`, `Playing Time_Min`, `Performance_CrdY`, `Team Success_PPM` | su versión `_rs` | la original |
| `Performance_Gls`, `Ast`, `Standard_Sh` | su versión `_rs` | la original y el `log1p_` |
| `Standard_SoT`, `Performance_Crs`, `Performance_Fls` | su `log1p_` | la original |

Regla simple: si hay `_rs` disponible, usamos esa (preserva mejor la forma de la distribución); si solo hay `log1p_`, usamos esa.

```python
cols_reemplazadas_por_rs = [
    "Age", "altura_m", "Playing Time_Min", "Performance_CrdY", "Team Success_PPM",
]
cols_reemplazadas_por_rs_y_log1p = ["Performance_Gls", "Performance_Ast", "Standard_Sh"]
cols_log1p_duplicado = [f"log1p_{c}" for c in cols_reemplazadas_por_rs_y_log1p]
cols_reemplazadas_por_log1p = ["Standard_SoT", "Performance_Crs", "Performance_Fls"]

df = df.drop(columns=(
    cols_reemplazadas_por_rs
    + cols_reemplazadas_por_rs_y_log1p
    + cols_log1p_duplicado
    + cols_reemplazadas_por_log1p
))
print(f"Shape actual: {df.shape}")
```

## 8. Categóricas: posición, liga y equipo

`posicion_tm` (13 categorías) y `Comp` (5 ligas) son de baja cardinalidad, así que van directo a one-hot. `Squad` (130 equipos) necesita otro tratamiento -- ver 8.3.

**Posición:** usamos `posicion_tm` (más granular) y tiramos `Pos` de FBref, que mezcla combinaciones para jugadores multiposición.

**Liga:** el EDA ya mostró que Premier League y La Liga tienen medianas de valor más altas -- se codifica completa.

**Equipo, por qué no one-hot:** 130 categorías, muchas con 1-2 observaciones. Agrupar en "top-N + otros" (lo que hacíamos antes) pierde la señal de equipos medianos como Betis o Villarreal, que sí influyen en el valor de sus jugadores. Evaluamos tres alternativas:

| Método | Problema |
|---|---|
| Label Encoding | Inventa un orden entre equipos que no existe |
| Frequency Encoding | Mide tamaño de plantilla, no prestigio |
| **Target Encoding** | Usa el target -- hay que calcularlo con cuidado para no filtrar información |

Target Encoding es el que mejor sirve, pero **no se calcula acá** para no meter fuga de información: `Squad` queda sin codificar en este notebook, y el encoding se ajusta en `04-modelado.ipynb` solo sobre el train, después del split por jugador.

```python
# 7.1 Posición: dummies desde posicion_tm, se elimina Pos (redundante, menos granular)
df = df.drop(columns=["Pos"])
dummies_pos = pd.get_dummies(df["posicion_tm"], prefix="pos", drop_first=True)

# 7.2 Competición: dummies desde Comp
dummies_comp = pd.get_dummies(df["Comp"], prefix="liga", drop_first=True)

print(f"Dummies de posición: {dummies_pos.shape[1]}")
print(f"Dummies de competición: {dummies_comp.shape[1]}")

df = pd.concat([df, dummies_pos, dummies_comp], axis=1)
df = df.drop(columns=["posicion_tm", "Comp"])

# 7.3 Equipo: se deja `Squad` como categórica sin codificar (130 categorías).
# El target encoding se ajusta en 04-modelado.ipynb solo con el train, tras el split.
df["Squad"] = df["Squad"].astype(str)
print(f"\nSquad se conserva sin codificar: {df['Squad'].nunique()} categorías (pendiente de target encoding en 04-modelado.ipynb)")

print(f"\nShape actual: {df.shape}")
```

## 9. Vencimiento de contrato: versión final

`contrato_vence` es una fecha, no sirve tal cual para los modelos. La señal ya vive en `dias_contrato_restante` (del EDA), así que tiramos la fecha cruda. Falta decidir qué hacer con su 43% de nulos, porque ni la regresión lineal ni Random Forest toleran `NaN`. Imputamos con la mediana global (no asumimos nada sobre esos jugadores) y dejamos `flag_contrato_conocido` para que el modelo pueda aprender si la ausencia del dato ya dice algo por sí sola.

```python
mediana_dias = df["dias_contrato_restante"].median()
df["dias_contrato_restante"] = df["dias_contrato_restante"].fillna(mediana_dias)
df = df.drop(columns=["contrato_vence"])

print(f"Mediana usada para imputar: {mediana_dias:.0f} días")
print(f"Nulos restantes en dias_contrato_restante: {df['dias_contrato_restante'].isna().sum()}")
print(f"Shape actual: {df.shape}")
```

## 10. Tipos finales y guardado

Convertimos las columnas booleanas a `0`/`1` y guardamos. `Squad` queda sin codificar a propósito (ver sección 8).

| Rol | Columnas |
|---|---|
| Identificadores | `Player`, `player_id`, `saison_id` |
| Objetivo | `valor_eur`, `log_valor_eur` (predecimos esta última) |
| Pendiente de codificar | `Squad` |
| Ya tratadas (`_rs`/`log1p_`) | edad, altura, minutos, goles, asistencias, tiros, amarillas, éxito de equipo, tiros a puerta, centros, faltas |
| Rendimiento sin redundancia | el resto de estadísticas crudas (penales, rojas, faltas recibidas, fuera de lugar, etc.) |
| Flags | sin tiros, sin titularidades, sin suplencias |
| Nuevas | `Ganador_Champions`, `Ganador_Liga` |
| Dummies | `pos_*`, `liga_*` |

Las variables de rendimiento que quedan sin tratar siguen correlacionando entre sí (el mismo cluster ofensivo/defensivo que vimos en el EDA), pero ninguna es duplicado exacto -- se recomienda revisar esto con regularización en el modelado.

```python
bool_cols = df.select_dtypes(include="bool").columns
df[bool_cols] = df[bool_cols].astype(int)

print(f"Shape final: {df.shape}")
print(f"Columnas booleanas convertidas a int: {len(bool_cols)}")
print()
print("Nulos restantes:", df.isna().sum().sum())
print()
df.dtypes.value_counts()
```
