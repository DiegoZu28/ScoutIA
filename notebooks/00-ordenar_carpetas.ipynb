```python
import shutil
import os
```

```python
ruta = '../data/raw/fbref_big5'
```

```python
for archivo in os.listdir(ruta):
    for i in range(1, 7):
        if f'202{i}-202{i+1}' in archivo and archivo.endswith('parquet'):
            shutil.move(os.path.join(ruta, archivo), os.path.join(ruta, f'202{i}-202{i+1}'))
        
```
